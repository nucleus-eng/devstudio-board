```python
%%writefile drive_utils.py
import os
import re

import gdown


def get_drive_file_id(url):
    """Extract the file ID from a Google Drive share URL."""
    match = re.search(r'/d/([a-zA-Z0-9_-]+)', url)
    if match:
        return match.group(1)
    return None


def download_from_drive(files_info, destination_dir):
    """
    Download files from Google Drive.

    Args:
        files_info (list): List of dicts with 'url' and 'filename'.
        destination_dir (str): Local path to save files.
    Returns:
        dict: Mapping of filenames to local paths.
    """
    os.makedirs(destination_dir, exist_ok=True)

    paths = {}
    for item in files_info:
        url = item['url']
        filename = item['filename']
        dest = os.path.join(destination_dir, filename)
        file_id = get_drive_file_id(url)

        if file_id:
            gdown.download(id=file_id, output=dest, quiet=True)
        else:
            gdown.download(url, output=dest, quiet=True)

        paths[filename] = dest
    return paths
```

## Plot fold-change comparison

```python
# --- steady-state fold change, per replicate well ---
def steady_state_values(df, name, n_last=10):
    tail = df.groupby(['Name', 'Well']).apply(lambda g: g.nlargest(n_last, 'Time_hrs')['Data'].mean())
    return tail[name].values  # one value per replicate well

def drop_lowest(values):
    values = list(values)
    values.remove(min(values))
    return values

def fold_change_per_well(df, base_name, ahsl_name, n_last=10, exclude_lowest_ahsl=False):
    base_vals = steady_state_values(df, base_name, n_last)
    ahsl_vals = steady_state_values(df, ahsl_name, n_last)
    if exclude_lowest_ahsl:
        ahsl_vals = drop_lowest(ahsl_vals)
    return ahsl_vals / base_vals.mean()

# --- compute for all eight conditions ---
fc_EsaO_01  = fold_change_per_well(df_1, "EsaO-mNG (0.1 nM)",    "EsaO-mNG (0.1 nM) + AHSL")
fc_EsaO2_01 = fold_change_per_well(df_2, "[EsaO]2-mNG (0.1 nM)", "[EsaO]2-mNG (0.1 nM) + AHSL")
fc_EsaO_05  = fold_change_per_well(df_3, "EsaO-mNG (0.5 nM)",    "EsaO-mNG (0.5 nM) + AHSL", exclude_lowest_ahsl=True)
fc_EsaO2_05 = fold_change_per_well(df_4, "[EsaO]2-mNG (0.5 nM)", "[EsaO]2-mNG (0.5 nM) + AHSL", exclude_lowest_ahsl=True)
fc_EsaO_1   = fold_change_per_well(df_5, "EsaO-mNG (1 nM)",      "EsaO-mNG (1 nM) + AHSL")
fc_EsaO2_1  = fold_change_per_well(df_6, "[EsaO]2-mNG (1 nM)",   "[EsaO]2-mNG (1 nM) + AHSL")
fc_EsaO_3   = fold_change_per_well(df_7, "EsaO-mNG (3 nM)",      "EsaO-mNG (3 nM) + AHSL")
fc_EsaO2_3  = fold_change_per_well(df_8, "[EsaO]2-mNG (3 nM)",   "[EsaO]2-mNG (3 nM) + AHSL")

# --- tidy into one long-form summary table, one row per replicate well ---
rows = []
for conc, construct, values in [
    (0.1, 'EsaO-mNG',    fc_EsaO_01),
    (0.1, '[EsaO]2-mNG', fc_EsaO2_01),
    (0.5, 'EsaO-mNG',    fc_EsaO_05),
    (0.5, '[EsaO]2-mNG', fc_EsaO2_05),
    (1,   'EsaO-mNG',    fc_EsaO_1),
    (1,   '[EsaO]2-mNG', fc_EsaO2_1),
    (3,   'EsaO-mNG',    fc_EsaO_3),
    (3,   '[EsaO]2-mNG', fc_EsaO2_3),
]:
    for v in values:
        rows.append({'DNA_conc_nM': conc, 'Construct': construct, 'Fold_change': v})

fold_summary = pd.DataFrame(rows)
print(fold_summary)
```

```python
#| label: cn-20260925-fig5

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10,5), sharey=True)

# --- Panel 1: EsaO-mNG ---
subset_1 = fold_summary[fold_summary['Construct'] == 'EsaO-mNG']
sns.barplot(data=subset_1, x='DNA_conc_nM', y='Fold_change', ax=ax1, color='tab:blue', errorbar='sd', capsize=0.1)
ax1.set(title='EsaO-mNG', xlabel='', ylabel='')

# --- Panel 2: [EsaO]2-mNG ---
subset_2 = fold_summary[fold_summary['Construct'] == '[EsaO]2-mNG']
sns.barplot(data=subset_2, x='DNA_conc_nM', y='Fold_change', ax=ax2, color='tab:orange', errorbar='sd', capsize=0.1)
ax2.set(title='[EsaO]2-mNG', xlabel='', ylabel='')

fig.supxlabel('DNA concentration (nM)')
fig.supylabel('RFU fold change (+AHSL / −AHSL)')

ax1.set_ylim(1, 2.5)

sns.despine(fig)
fig.tight_layout()
plt.show()
```

## Kinetic Analysis

See the [DevNote](https://devnotes.nucleus.engineering/articles/Newman-20260421) on kinetic analysis for more details.

**Metrics extracted:**

  - **Maximum velocity**: Maximum rate of fluorescence increase (slope at inflection point)
  - **Lag time**: Time to reach the exponential phase
  - **Steady-state**: Final fluorescence level
  - **Time-to-completion**: Time it takes to reach 95% of asymptote
  - **Drift**: Rate of signal decay or increase after steady-state
  - **R²**: Goodness of fit; "Good Fit" is `True` if $R^2 \geq 0.95$

## Key Metrics Explained

### 1. **Steady-State Level** (`Steady State, Data`)
- The final fluorescence value reached by the reaction
- Represents the total amount of protein produced
- Higher values indicate greater expression yield

### 2. **Maximum Velocity** (`Velocity, Max`)
- The steepest slope of the fluorescence curve (at the inflection point)
- Units: RFU per second
- Reflects the peak rate of protein synthesis
- Sensitive to enzyme activity, substrate availability, and reaction conditions

### 3. **Lag Time** (`Lag, Time`)
- Time before exponential fluorescence increase begins
- May reflect time for ribosome assembly or initial translation steps
- Shorter lag times suggest faster reaction initiation

### 4. **Drift** (`Fit, drift`)
- Rate of fluorescence change after reaching steady-state
- Positive drift: continued synthesis or aggregation
- Negative drift: photobleaching, protein degradation, or quenching
- Units: RFU per second

### 5. **R² Value** (`Fit, R^2`)
- Goodness of fit (0 to 1, higher is better)
- R² > 0.98 indicates excellent fit
- Poor fits may indicate noisy data, overflow errors, or non-sigmoid kinetics

---

## Tips and Troubleshooting

- **Overflow errors:** Wells with `OVRFLW` or `NaN` values are automatically excluded from fitting
- **Poor fits (low R²):** Inspect raw curves for anomalies (bubbles, evaporation, pipetting errors)
- **Drift:** Sometimes seen in kinetics curves; use `sigmoid_drift` model
- **Multiple replicates:** Always include technical replicates and report error bars
- **Comparing conditions:** Normalize or blank data consistently across all samples
